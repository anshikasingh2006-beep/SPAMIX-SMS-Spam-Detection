# Notebook 3: Traditional NLP Baselines (Naive Bayes & Logistic Regression)

### Objective:
1. Construct TF-IDF vectorizer (n-grams 1-2, sublinear scaling).
2. Train Multinomial Naive Bayes.
3. Train Logistic Regression.
4. Evaluate classification metrics, confusion matrices, and inference latency.

```python
import os, joblib
import pandas as pd
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
```

## 1. Load Data & Serialized Baselines

```python
test_df = pd.read_csv('../data/processed/test.csv')
y_test = (test_df['label'] == 'spam').astype(int).values

tfidf = joblib.load('../models/tfidf_vectorizer.joblib')
nb = joblib.load('../models/naive_bayes_model.joblib')
lr = joblib.load('../models/logistic_regression_model.joblib')

X_test_tfidf = tfidf.transform(test_df['message'])
print('TF-IDF Test Shape:', X_test_tfidf.shape)
```

## 2. Naive Bayes Evaluation

```python
nb_preds = nb.predict(X_test_tfidf)
print('--- Multinomial Naive Bayes ---')
print(classification_report(y_test, nb_preds, target_names=['Ham', 'Spam']))
```

## 3. Logistic Regression Evaluation

```python
lr_preds = lr.predict(X_test_tfidf)
print('--- Logistic Regression ---')
print(classification_report(y_test, lr_preds, target_names=['Ham', 'Spam']))
```
