# Notebook 2: Exploratory Data Analysis & Feature Engineering

### Objective:
1. Analyze message length distributions (character & word level).
2. Extract structural and syntactic features (digits, uppercase ratio, URLs, currency hooks).
3. Investigate top unigrams and bigrams for spam vs ham.

```python
import os, sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append('..')
from src.feature_engineering import add_feature_columns
```

## 1. Feature Extraction

```python
df = pd.read_csv('../data/processed/sms_spam_clean.csv')
df_features = add_feature_columns(df)
print('Features Shape:', df_features.shape)
df_features.head()
```

## 2. Character Length Analysis

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df_features, x='char_count', hue='label', bins=60, kde=True, palette={'ham':'#22c55e', 'spam':'#ef4444'})
plt.title('Character Count Distribution: Ham vs Spam')
plt.xlim(0, 300)
plt.show()
```

## 3. Structural Signal Analysis (Uppercase & Digits)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df_features, x='label', y='digit_count', ax=axes[0], palette={'ham':'#22c55e', 'spam':'#ef4444'})
axes[0].set_title('Digit Count by Message Class')
axes[0].set_ylim(0, 25)

sns.boxplot(data=df_features, x='label', y='caps_ratio', ax=axes[1], palette={'ham':'#22c55e', 'spam':'#ef4444'})
axes[1].set_title('Uppercase Ratio by Message Class')
plt.tight_layout()
plt.show()
```
